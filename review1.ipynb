# Q1. Simple Caesar Shift

##### A basic encryption exercise requires shifting each letter in a word forward by 1 in the alphabet(z wraps to a) sample input

```python
word="cab"
```

```python
encrypted=""
for ch in word:
    if ch=="z":
        encrypted+="a"
    else:
        encrypted += chr(ord(ch)+1)
print("shifted word:",encrypted)

        
```

# Q2.Internet Speed Test Logger

##### An ISP monitors connection speeds.classify each test as 'Excellent'(>=100 Mbps),'good'(50-99),'fair'(20-49),'poor'(<20).

```python
speed_tests=[105,45,18,78,92,15,60,110,33,8]
```

```python
Excellent=0
Good=0
Fair=0
Poor=0
```

```python
for speed in speed_tests:
    if speed>=100:
        Excellent+=1
        
    elif speed>=50:
        Good+=1
        
    elif speed>=20:
        Fair+=1
    
    else:
         Poor+=1
print("Excellent:",Excellent)
print("Good:",Good)
print("Fair:",Fair)
print("Poor:",Poor)
        
```

```python
average=sum(speed_tests)/len(speed_tests)
print("average:",average)
```
